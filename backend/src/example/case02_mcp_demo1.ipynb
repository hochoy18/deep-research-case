```python
import json

import dotenv
from dashscope import Application
import os
from rich import print as rprint
from agent.base_agent import WebSearchAgent
from dashscope.app.application_response import ApplicationResponse


dotenv.load_dotenv(override=True)
api_key = os.getenv("MCP_APP_TOKEN")
app_id = os.getenv("MCP_APP_ID")
```

```python

web_searcher = WebSearchAgent()

search_topic = '2025 2026 企业级 AI Agent 框架平台对比 LangChain AutoGen CrewAI Salesforce Agentforce Microsoft Copilot Studio 落地趋势'
# 执行搜索
response:ApplicationResponse = Application.call(
          api_key=api_key,
          app_id=app_id,
          prompt = search_topic,
          biz_params={'count':5}
      )
# rprint(response)

# response = web_searcher.step(prompt=search_topic,
#                              count=10)
```
