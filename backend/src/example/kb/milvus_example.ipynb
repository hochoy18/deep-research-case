```python
import os

import dotenv
from pymilvus import MilvusClient
from agent.exceptions import (
    KBConnectionError,
    KBEmbeddingError,
    KBEmbeddingFatalError,
    KBConfigError,
)
from pymilvus.milvus_client.index import IndexParams
```

```python
milvus_uri = 'http://localhost:19530'
collection_name = 'deep_research_case'
COLLECTION_NAME = "research_facts"

DEFAULT_EMBEDDING_DIM = 1024
# DEFAULT_EMBEDDING_MODEL = "text-embedding-v3"

DEFAULT_EMBEDDING_MODEL = 'qwen3.7-text-embedding'
```

```python
import dashscope
from http import HTTPStatus
input_texts = "衣服的质量杠杠的，很漂亮，不枉我等了这么久啊，喜欢，以后还来这里买"

dotenv.load_dotenv(override=True)
EMBEDDING_API_KEY = os.getenv('EMBEDDING_API_KEY')
resp = dashscope.TextEmbedding.call(
    api_key=EMBEDDING_API_KEY,
    model=DEFAULT_EMBEDDING_MODEL,
    input=input_texts
)
print(resp)
```

```python
client = MilvusClient(uri=milvus_uri)

```

```python
collections = client.list_collections()
```

```python
from typing import List

text = '研究欧美酒店旅游行业 AI Agent 在企业如何落地'


resp = dashscope.TextEmbedding.call(
    api_key=EMBEDDING_API_KEY,
    model=DEFAULT_EMBEDDING_MODEL,
    input=text
)
print(resp)

```

```python

search_all: List[List[dict]] = client.search(
    collection_name=COLLECTION_NAME,
    data=[resp.output.get('embeddings',[])[0].get('embedding',[])],

     limit=20,
    output_fields=[
        "fact_text", "source_url", "research_topic",
        "confidence", "created_at", "fact_category",
    ],

)


print(len(search_all))
count = 10
for item in search_all[:count]:
    print(item)
```
